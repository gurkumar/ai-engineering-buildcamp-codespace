```python
messages=[
    {
        "role":"user",
        "content":"Tell me about a bed tiem stort about unicorn"
    }
]
responses = openai_client.responses.create(
    model="gpt-4o-mini",
    input=messages
)
print(responses.output_text)
```

```python
responses.usage
```

```python
MODEL_PRICES={ # dictionary for model prices
    "gpt-4o-mini":{"input":0.15,"output":0.60},
    "gpt-5-nano":{"input":0.075,"output":0.30}, # cheapest
    "gpt-5-mini":{"input":0.25,"output":2.00},
    "gpt-5.2":{"input":1.75,"output":14.00},
    "gpt-5.2-pro":{"input":21.00,"output":168.00}, # most expensive
}
```

```python
def calculate_cost(model_name:str, input_tokens:int,output_tokens:int)->float:
    prices = MODEL_PRICES[model_name.lower()]
    input_cost = (input_tokens/1_000_000) * prices["input"] # price per 1M tokens
    output_cost=(output_tokens/1_000_000) * prices["output"]
    return input_cost + output_cost
```

```python
usage = responses.usage
cost = calculate_cost(
    model_name="gpt-4o-mini",
    input_tokens=usage.input_tokens,
    output_tokens=usage.output_tokens
)
print(f"Cost: ${cost:.6f}") # 6 floating point digist after decimal
```

```python
#instead of waiting for output as it can be long we can stream
messages=[
    {
        "role":"user",
        "content":"Tell me about a bed tiem stort about unicorn"
    }
]
stream = openai_client.responses.create(
    model="gpt-4o-mini",
    input=messages,
    stream=True
)

events=[]
for event in stream:
    events.append(event)
    if hasattr(event,'delta'): # delta  is piece of text we want to present
        print(event.delta,end='')
```

```python
#instead of waiting for output as it can be long we can stream
messages=[
    {
        "role":"user",
        "content":"Tell me about a bed tiem stort about unicorn"
    }
]
stream = openai_client.responses.create(
    model="gpt-4o-mini",
    input=messages,
    stream=True
)

response=None
for event in stream:
    
    if hasattr(event,'delta'): # delta  is piece of text we want to present
        print(event.delta,end='')
    if hasattr(event,'response'):
        response = event.response

print(response.usage)
```
