```python
import requests
HEADERS = {"User-Agent": "ai-engineering-buildcamp/1.0 (educational project)"}
def search_wikipedia(query:str) -> list[dict]:
    """ Search Wikipedia for pages related to query.

    Args:
       query: search terms, e.g. capybara

    Returns:
       List of search results (each has title, page id, snippet, size, wordcount,timestamp).
    """
    url="https://en.wikipedia.org/w/api.php"
    #https://en.wikipedia.org/w/api.php?action=query&format=json&list=search&srsearch=YOUR_QUERY
    params={
        "action":"query",
        "format":"json",
        "list":"search",
        "srsearch":query,

    }
  
    response = requests.get(url, params=params, headers=HEADERS, timeout=30)
    response.raise_for_status()
    return response.json()
```

```python
results = search_wikipedia("capybara")
print(len(results["query"]["search"]))
```
